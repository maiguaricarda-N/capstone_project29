Is GDP group related to child mortality group?

```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
```

```python

# Load data
df = pd.read_csv(r"C:\Users\admin\Desktop\Captoneproject-pydata\data\clean_joined.csv")

# Check the data
print(df.shape)
df.head()
```

```python
#  Checking the two groups
print(df["gdp_group"].value_counts())
print(df["mort_group"].value_counts())
```

```python
#  writing my hypotheses before I run the test
print("H0: GDP group and child mortality group are independent (not related).")
print("H1: GDP group and child mortality group are related.")
print("I will reject H0 if the p-value is below 0.05.")
```

```python
#Creating the contingency table
table = pd.crosstab(df["gdp_group"], df["mort_group"])

print(table)
```

```python
# running the chi square        
chi2, p_value, degrees, expected = stats.chi2_contingency(table)

print("Chi-square:", round(chi2, 3))
print("p-value:", round(p_value, 5))
print("p-value:", "less than 0.001" if p_value < 0.001 else round(p_value, 3))
```

```python
# interpreting the quiz
if p_value < 0.05:
    print("Reject H0.")
    print("There is evidence of a relationship between GDP group and child mortality group.")
else:
    print("Fail to reject H0.")
    print("There is not enough evidence of a relationship between GDP group and child mortality group.")
```
